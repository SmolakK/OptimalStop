```python
import pickle, pandas as pd, numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

with open('results_dbscan.pkl','rb') as f:
    results = pickle.load(f)

def parse_key(k, keys=('eps','min_samples','stop_distance','stop_time')):
    vals = eval(k)                       # "[50, 3, 30, '5T']"  →  [50,3,30,'5T']
    return dict(zip(keys, vals))

rows = []
for k, df in results.items():           # 200 parameter sets
    pars = parse_key(k)
    for person_id, metrics in df.iterrows():
        row = metrics.copy()
        row['cfg_key']  = k
        row['person']   = person_id
        for p, v in pars.items():
            row[p] = v
        rows.append(row)

big = pd.DataFrame(rows)
```

```python
def extract_pareto(df, x='Pred', y='Real', group='person'):
    """
    Returns a DataFrame with Pareto-optimal points for each person,
    maximising both x and y.
    """
    def pareto_group(g):
        g_sorted = g.sort_values([x, y], ascending=[False, False])
        best_y = -float('inf')
        frontier = []
        for _, row in g_sorted.iterrows():
            if row[y] > best_y:
                frontier.append(row)
                best_y = row[y]
        return pd.DataFrame(frontier)

    return df.groupby(group, group_keys=False).apply(pareto_group).reset_index(drop=True)

pareto_df = extract_pareto(big)
```
