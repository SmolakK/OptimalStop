```python
import pandas as pd
import numpy as np
import geopandas as gpd
from tqdm import tqdm
import os
import pickle

def load_pickle_dict(path: str) -> dict:
    with open(path, 'rb') as f:
        data = pd.read_pickle(f)
    return data
```
