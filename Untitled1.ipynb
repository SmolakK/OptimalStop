```python
cols = X.columns
indi = X.index
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled,columns=cols,index=indi)
```

```python
split = .7
users = df.index.get_level_values(0).unique()
split_point = ceil(len(users)*split)
users_train = np.random.choice(users,split_point)
users_test = [x for x in users if x not in users_train]
```

```python
X_train = X_scaled.loc[users_train]
y_train = y.loc[users_train]
X_test = X_scaled.loc[users_test]
y_test = y.loc[users_test]
```

```python
X_train['person'] = X_train.index
X_train = X_train.reset_index(drop=True)
X_test['person'] = X_test.index
X_test = X_test.reset_index(drop=True)
```

```python
import xgboost as xgb
import optuna
from sklearn.model_selection import GroupKFold, cross_val_score
import numpy as np

# Prepare data once, outside of Optuna
groups = X_train['person'].values
X_train_clean = X_train.drop(columns=['person'])
X_test_clean = X_test.drop(columns=['person'])

def objective(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 1000),
        'max_depth': trial.suggest_int('max_depth', 3, 15),
        'learning_rate': trial.suggest_float('learning_rate', 1e-3, 0.3, log=True),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
        'gamma': trial.suggest_float('gamma', 0, 5),
        'min_child_weight': trial.suggest_int('min_child_weight', 1, 20),
        'random_state': 42,
        'objective': 'reg:squarederror'
    }

    model = xgb.XGBRegressor(**params)
    cv = GroupKFold(n_splits=5)
    scores = cross_val_score(model, X_train_clean, y_train, groups=groups, cv=cv,
                             scoring='neg_mean_absolute_error', n_jobs=-1)
    return -np.mean(scores)

# Run optimization
study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=50)

# Print results
print("Best parameters:", study.best_params)
print("Best CV MAE:", study.best_value)
```

```python
study.best_params
```

```python
best_params = study.best_params
print("Best parameters:", best_params)

best_model = xgb.XGBRegressor(**best_params)
best_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_absolute_error
y_pred = best_model.predict(X_test)
print("Test MAE:", mean_absolute_error(y_test, y_pred))
```
