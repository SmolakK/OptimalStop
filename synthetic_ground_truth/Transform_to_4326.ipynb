```python
import pandas as pd
from pyproj import Transformer
df = pd.read_csv(r'D:\GitHub\OptimalStop\synthetic_ground_truth\random_city.csv').iloc[:, 1:]
```

```python
df
```

```python
transformer = Transformer.from_crs("epsg:3857", "epsg:4326", always_xy=True)
df[['lon', 'lat']] = df.apply(
    lambda row: transformer.transform(row['lon'], row['lat']), axis=1,
    result_type='expand')
```

```python
df.to_csv(r'D:\GitHub\OptimalStop\synthetic_ground_truth\random_city_4326.csv')
```
